### What makes people say yes to a second date?

```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go
import plotly.express as px
import chardet 
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

```python
file = 'Speed+Dating+Data.csv'

#detect file encoding

with open(file, 'rb') as f:
    data = f.read()

# Detect the encoding
result = chardet.detect(data)
print(result['encoding'])

# Build the Dataframe
data = pd.read_csv(file, encoding=result['encoding'])
df = pd.DataFrame(data)

df.sample(30)
```

```python
df.describe(include='all')
```

```python
df.columns
```

# Identify unique member from each wave

```python
unique = df[['iid', 'gender', 'age', 'race', 'goal', 'date', 'go_out']].drop_duplicates('iid')

unique['gender_label'] = unique['gender'].map({0: 'female', 1: 'male'})

race_map = {1: 'Black/African American', 2: 'European/Caucasian', 3: 'Latino/Hispanic',
            4: 'Asian/Pacific Islander', 5: 'Native American', 6: 'Other'}
unique['race_label'] = unique['race'].map(race_map)

goal_map = {1: 'Fun night out', 2: 'Meet new people', 3: 'Get a date',
            4: 'Serious relationship', 5: 'Say I did it', 6: 'Other'}
unique['goal_label'] = unique['goal'].map(goal_map)

unique
```

```python
print(f'Total participants: {len(unique)}')
print(f'Total speed dates recorded: {len(df)}')
print(f'Overall match rate: {df["match"].mean()*100:.1f}%')
```

# 1. Sample Overview

### 1.1. Gender distribution

```python
counts = unique['gender_label'].value_counts().reset_index()
counts.columns = ['Gender', 'Count']

fig = px.pie(
    counts, values='Count', names='Gender',
    title='Gender Distribution of Participants',
    color='Gender',
    hole=0.5
)
 
fig.update_traces(
    texttemplate='%{label}<br>%{value} (%{percent})',
    textposition='outside'
)
 
fig.update_layout(
    showlegend=False,
    title_font_size=18,
    margin=dict(t=60, b=40, l=40, r=40)
)
 
fig.show()

```

**Caption:** The dataset is nearly perfectly balanced with 274 women (49.7%) and 277 men (50.3%). This balance is by design — each speed dating event was organized with equal numbers of men and women. For our analysis of second date decisions, this means we can compare male and female behavior without worrying about sampling bias.

### 1.2 Age Distribution

```python
f_mean, f_std = unique[unique['gender']==0]['age'].mean(), unique[unique['gender']==0]['age'].std()
m_mean, m_std = unique[unique['gender']==1]['age'].mean(), unique[unique['gender']==1]['age'].std()
```

```python
fig = px.histogram(
    unique, x='age',facet_col='gender_label',
    nbins=20,
    color='gender_label',
    title='Age Distribution by Gender — Mean and ±1 Standard Deviation',
    labels={'age': 'Age', 'gender_label': 'Gender'}
)

for col, mean, std, color in [(1, f_mean, f_std, 'Black'),(2, m_mean, m_std, 'Black')]:
    fig.add_vline(x=mean, line_dash='solid', line_color=color, line_width=2,
                  annotation_text=f'Mean = {mean:.1f}',
                  annotation_position='top right',
                  annotation_font_color=color, col=col)
    fig.add_vline(x=mean - std, line_dash='dash', line_color=color, line_width=1.5,
                  annotation_text=f'-1 SD ({mean-std:.1f})',
                  annotation_position='bottom left',
                  annotation_font_color=color, col=col)
    fig.add_vline(x=mean + std, line_dash='dash', line_color=color, line_width=1.5,
                  annotation_text=f'+1 SD ({mean+std:.1f})',
                  annotation_position='bottom right',
                  annotation_font_color=color, col=col)    

fig.update_layout(showlegend=False)

fig.show()
```

**Caption:** Most participants are between 22 and 30 years old, reflecting the graduate student population at Columbia University. 

### 1.3 Race Distribution

```python
counts_race = unique['race_label'].value_counts().reset_index()
counts_race.columns = ['Race', 'Count']

fig = px.bar(counts_race,
             x='Count', y='Race',
             title='Racial Composition of Participants',
             color='Count',
             color_continuous_scale='Blues',
             text='Count',             
)

fig.update_traces(textposition='outside')
fig.update_layout(
    coloraxis_showscale=False,
    yaxis=dict(categoryorder='total ascending'),
    xaxis_title='Number of participants',
    yaxis_title='',
    title_font_size=16
)

fig.show()
```

European/Caucasian participants are the largest group (304 out of 545 with valid data), followed by Asian/Pacific Islander (136). The sample reflects the demographics of Columbia University's graduate programs in the early 2000s. 

### 1.4 Participant Goals

```python
counts_goal = unique['goal_label'].value_counts().reset_index()
counts_goal.columns = ['Goal', 'Count']
counts_goal['Percentage'] = (counts_goal['Count'] / counts_goal['Count'].sum() * 100).round(1)

fig = px.bar(
    counts_goal, x='Goal', y='Count',
    title='Why did participants sign up?',
    color='Count',
    color_continuous_scale='Pinkyl',
    text=counts_goal['Percentage'].apply(lambda x: f'{x}%')
)
fig.update_traces(textposition='inside')
fig.update_layout(
    coloraxis_showscale=False,
    xaxis_title='', yaxis_title='Number of participants',
    title_font_size=16
)
fig.show()
```
