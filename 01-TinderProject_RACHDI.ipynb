```python
import pandas as pd
import numpy as np
import chardet 
```

```python
file = 'Speed+Dating+Data.csv'
```

```python
# detect file encoding

with open(file, 'rb') as f:
    data = f.read()

# Detect the encoding
result = chardet.detect(data)
print(result['encoding'])
```

```python
data = pd.read_csv(file, encoding=result['encoding'])
df = pd.DataFrame(data)
```
